```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

# Load data
u_les = np.load("u_les.npy")
v_les = np.load("v_les.npy")
u_nudging = np.load("u_nudging.npy")
v_nudging = np.load("v_nudging.npy")
u_dif = np.load("u_dif.npy")
v_dif = np.load("v_dif.npy")

# Compute magnitude of differences
mag_dif = np.sqrt(u_dif**2 + v_dif**2)

# Grid setup
nt, ny, nx = u_les.shape
Lx, Ly = 2.0, 2.0
x = np.linspace(0, Lx, nx)
y = np.linspace(0, Ly, ny)
X, Y = np.meshgrid(x, y)

# Normalize for color range
vmin, vmax = mag_dif.min(), mag_dif.max()

# Create the figure and axis
fig, ax = plt.subplots(figsize=(6, 5))

def update(frame):
    ax.clear()
    ax.contourf(X, Y, mag_dif[frame], levels=50, cmap='magma', vmin=vmin, vmax=vmax)
    ax.set_title(f"Magnitude of Difference - Frame {frame}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    return []

ani = animation.FuncAnimation(fig, update, frames=range(0, nt, 10), blit=False)
plt.close()
HTML(ani.to_jshtml())
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

# Simulation grid
nx, ny = 100, 100
nt = 100
Lx, Ly = 2.0, 2.0
x = np.linspace(0, Lx, nx)
y = np.linspace(0, Ly, ny)
X, Y = np.meshgrid(x, y)

# Load precomputed velocity fields
u_les = np.load("u_les_100.npy")
v_les = np.load("v_les_100.npy")
u_nudging = np.load("u_nudging_100.npy")
v_nudging = np.load("v_nudging_100.npy")

# Compute differences
u_dif = u_les - u_nudging
v_dif = v_les - v_nudging
mag_dif = np.sqrt(u_dif**2 + v_dif**2)
```

```python
fig1, ax1 = plt.subplots()
contour1 = ax1.contourf(X, Y, np.sqrt(u_les[0]**2 + v_les[0]**2), levels=50, cmap='viridis')
title1 = ax1.set_title("LES Velocity Magnitude")

def update_les(n):
    for c in contour1.collections:
        c.remove()
    contour = ax1.contourf(X, Y, np.sqrt(u_les[n]**2 + v_les[n]**2), levels=50, cmap='viridis')
    title1.set_text(f"LES Velocity Magnitude (t={n})")
    return contour.collections

ani1 = animation.FuncAnimation(fig1, update_les, frames=range(0, nt, 10), blit=False)
plt.close(fig1)
HTML(ani1.to_jshtml())
```

```python
fig2, ax2 = plt.subplots()
contour2 = ax2.contourf(X, Y, np.sqrt(u_nudging[0]**2 + v_nudging[0]**2), levels=50, cmap='plasma')
title2 = ax2.set_title("Nudging Velocity Magnitude")

def update_nudging(n):
    for c in contour2.collections:
        c.remove()
    contour = ax2.contourf(X, Y, np.sqrt(u_nudging[n]**2 + v_nudging[n]**2), levels=50, cmap='plasma')
    title2.set_text(f"Nudging Velocity Magnitude (t={n})")
    return contour.collections

ani2 = animation.FuncAnimation(fig2, update_nudging, frames=range(0, nt, 10), blit=False)
plt.close(fig2)
HTML(ani2.to_jshtml())
```

```python
fig3, ax3 = plt.subplots()
contour3 = ax3.contourf(X, Y, mag_dif[0], levels=50, cmap='magma')
title3 = ax3.set_title("Difference Magnitude")

def update_difference(n):
    for c in contour3.collections:
        c.remove()
    contour = ax3.contourf(X, Y, mag_dif[n], levels=50, cmap='magma')
    title3.set_text(f"Difference Magnitude (t={n})")
    return contour.collections

ani3 = animation.FuncAnimation(fig3, update_difference, frames=range(0, nt, 10), blit=False)
plt.close(fig3)
HTML(ani3.to_jshtml())
```
