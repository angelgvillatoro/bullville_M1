```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# =============================================================================
# PARÁMETROS Y CONSTANTES
# =============================================================================
# Propiedades del aire (25°C, 1 atm)
rho = 1.184       # Densidad [kg/m³]
mu = 1.849e-5     # Viscosidad dinámica [Pa·s]
nu = mu/rho       # Viscosidad cinemática [m²/s]
k = 0.0262        # Conductividad térmica [W/(m·K)]
alpha = 2.22e-5   # Difusividad térmica [m²/s]
cp = 1006.0       # Calor específico [J/(kg·K)]

# Escalas características
L = 1.0           # Longitud característica [m]
H = 0.1           # Altura característica [m]
u0 = 2.0          # Velocidad característica [m/s]
up = -1.0         # Velocidad placa [m/s]
T0 = 293.0        # Temperatura referencia [K]
p0 = 101325.0     # Presión referencia [Pa]

# Cálculo de u_c, p_f y T_f
uc = u0 - up      # Velocidad característica corregida
pf = mu*uc*L/H**2 + p0
Tf = T0 + (mu*uc**2)/k

# Parámetros adimensionales
Re = 20.0                     # Reynolds
Pr = 10.0                     # Prandtl
Ec = 0.1                      # Eckert
Eu = 1.0                      # Euler (presión)

# =============================================
# PARÁMETROS ADIMENSIONALES
# =============================================
Lx_star, Ly_star = 1.0, 1.0  # Dimensiones adimensionales
nx, ny = 50, 50               # Puntos en la malla
dx_star = Lx_star/(nx-1)      # Paso espacial x
dy_star = Ly_star/(ny-1)      # Paso espacial y

# Condiciones de frontera
up = -0.5               # Velocidad placa
u0 = 1.0                # Velocidad flujo libre
T0_star = 1.0           # Temperatura pared inferior
T1_star = 0.0           # Temperatura pared superior

# Tiempo
dt_star = 0.001         # Paso temporal
nt = 500                # Pasos totales

# =============================================
# INICIALIZACIÓN
# =============================================
x_star = np.linspace(0, Lx_star, nx)
y_star = np.linspace(0, Ly_star, ny)
X_star, Y_star = np.meshgrid(x_star, y_star)

# Campos iniciales
u_star = np.zeros((ny, nx)) + u0
v_star = np.zeros((ny, nx))
p_star = np.zeros((ny, nx))
T_star = np.zeros((ny, nx)) + T1_star

# Condiciones de frontera
u_star[0, :] = up       # Placa inferior móvil
u_star[-1, :] = u0      # Flujo libre superior
v_star[0, :] = 0        # No penetración inferior
v_star[-1, :] = 0       # No penetración superior
T_star[0, :] = T0_star  # Temperatura pared inferior
T_star[-1, :] = T1_star # Temperatura pared superior

# =============================================
# SIMULACIÓN
# =============================================
u_history, v_history, p_history, T_history, tau_history = [], [], [], [], []

for n in range(nt):
    u_old, v_old, p_old, T_old = u_star.copy(), v_star.copy(), p_star.copy(), T_star.copy()
    
    # Primero resolvemos las ecuaciones de momento
    for i in range(1, ny-1):
        for j in range(1, nx-1):
            # Términos convectivos
            conv_u_x = u_old[i,j] * (u_old[i,j+1] - u_old[i,j-1])/(2*dx_star)
            conv_u_y = v_old[i,j] * (u_old[i+1,j] - u_old[i-1,j])/(2*dy_star)
            
            # Términos difusivos
            diff_u_x = (u_old[i,j+1] - 2*u_old[i,j] + u_old[i,j-1])/(dx_star**2)
            diff_u_y = (u_old[i+1,j] - 2*u_old[i,j] + u_old[i-1,j])/(dy_star**2)
            
            # Gradiente de presión
            # grad_p_x = (p_old[i,j+1] - p_old[i,j-1])/(2*dx_star)
            grad_p_x = 0
            
            # Ecuación de momento x (actualizada)
            u_star[i,j] = u_old[i,j] + dt_star*(-conv_u_x - conv_u_y - Eu*grad_p_x + (1/Re)*(diff_u_x + diff_u_y))
            
            # Ecuación de momento y
            conv_v_x = u_old[i,j] * (v_old[i,j+1] - v_old[i,j-1])/(2*dx_star)
            conv_v_y = v_old[i,j] * (v_old[i+1,j] - v_old[i-1,j])/(2*dy_star)
            
            diff_v_x = (v_old[i,j+1] - 2*v_old[i,j] + v_old[i,j-1])/(dx_star**2)
            diff_v_y = (v_old[i+1,j] - 2*v_old[i,j] + v_old[i-1,j])/(dy_star**2)
            
            grad_p_y = (p_old[i+1,j] - p_old[i-1,j])/(2*dy_star)
            
            v_star[i,j] = v_old[i,j] + dt_star*(-conv_v_x - conv_v_y - Eu*grad_p_y + (1/Re)*(diff_v_x + diff_v_y))
    
    # Resolver ecuación de continuidad para presión (método de proyección)
    for _ in range(20):  # Iteraciones para presión
        for i in range(1, ny-1):
            for j in range(1, nx-1):
                p_star[i,j] = 0.25*(p_old[i+1,j] + p_old[i-1,j] + p_old[i,j+1] + p_old[i,j-1] - 
                              (dx_star*dy_star)/(2*(dx_star**2 + dy_star**2)) * (
                                  (u_star[i,j+1] - u_star[i,j-1])/(2*dx_star) + 
                                  (v_star[i+1,j] - v_star[i-1,j])/(2*dy_star)))
    
    # Actualizar velocidades con nuevo gradiente de presión
    for i in range(1, ny-1):
        for j in range(1, nx-1):
            u_star[i,j] -= dt_star * Eu * (p_star[i,j+1] - p_star[i,j-1])/(2*dx_star)
            v_star[i,j] -= dt_star * Eu * (p_star[i+1,j] - p_star[i-1,j])/(2*dy_star)
    
    # Resolver ecuación de energía
    for i in range(1, ny-1):
        for j in range(1, nx-1):
            # Términos convectivos
            conv_T_x = u_star[i,j] * (T_old[i,j+1] - T_old[i,j-1])/(2*dx_star)
            conv_T_y = v_star[i,j] * (T_old[i+1,j] - T_old[i-1,j])/(2*dy_star)
            
            # Términos difusivos
            diff_T_x = (T_old[i,j+1] - 2*T_old[i,j] + T_old[i,j-1])/(dx_star**2)
            diff_T_y = (T_old[i+1,j] - 2*T_old[i,j] + T_old[i-1,j])/(dy_star**2)
            
            # Disipación viscosa
            Sxx = (u_star[i,j+1] - u_star[i,j-1])/(2*dx_star)
            Syy = (v_star[i+1,j] - v_star[i-1,j])/(2*dy_star)
            Sxy = 0.5*((u_star[i+1,j] - u_star[i-1,j])/(2*dy_star) + 
                       (v_star[i,j+1] - v_star[i,j-1])/(2*dx_star))
            
            viscous_heating = (Ec/Re)*(2*(Sxx**2 + Syy**2) + 4*Sxy**2)
            
            # Ecuación de energía
            T_star[i,j] = T_old[i,j] + dt_star*(-conv_T_x - conv_T_y + 
                                                (1/(Re*Pr))*(diff_T_x + diff_T_y) + 
                                                viscous_heating)
    
    # Calcular shear stress en todo el dominio (τ = μ(du/dy))
    tau = np.zeros((ny, nx))
    for i in range(1, ny-1):
        for j in range(nx):
            tau[i,j] = (u_star[i+1,j] - u_star[i-1,j])/(2*dy_star)  # du/dy
    
    # Condiciones de frontera
    v_star[:, 0], v_star[:, -1] = 0, 0  # No penetración en paredes laterales
    u_star[0, :], u_star[-1, :] = up, u0
    v_star[0, :], v_star[-1, :] = 0, 0
    T_star[0, :], T_star[-1, :] = T0_star, T1_star
    
    # Condiciones de frontera para presión (Neumann)
    p_star[:, 0] = p_star[:, 1]    # ∂p/∂x = 0 en x=0
    p_star[:, -1] = p_star[:, -2]  # ∂p/∂x = 0 en x=Lx
    p_star[0, :] = p_star[1, :]    # ∂p/∂y = 0 en y=0
    p_star[-1, :] = p_star[-2, :]  # ∂p/∂y = 0 en y=Ly
    
    if n % 10 == 0:
        u_history.append(u_star.copy())
        v_history.append(v_star.copy())
        p_history.append(p_star.copy())
        T_history.append(T_star.copy())
        tau_history.append(tau.copy())

# =============================================
# VISUALIZACIÓN ANIMADA
# =============================================
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))

# Configuración inicial
speed = np.sqrt(u_history[0]**2 + v_history[0]**2)
cont1 = ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
cbar1 = fig.colorbar(cont1, ax=ax1)
cbar1.set_label('|u*|')
title1 = ax1.set_title(f'Velocidad (t*=0.00)')
ax1.set_xlabel('x*')
ax1.set_ylabel('y*')

cont2 = ax2.contourf(X_star, Y_star, T_history[0], levels=20, cmap='jet')
cbar2 = fig.colorbar(cont2, ax=ax2)
cbar2.set_label('T*')
title2 = ax2.set_title('Temperatura adimensional')
ax2.set_xlabel('x*')
ax2.set_ylabel('y*')

cont3 = ax3.contourf(X_star, Y_star, p_history[0], levels=20, cmap='viridis')
cbar3 = fig.colorbar(cont3, ax=ax3)
cbar3.set_label('p*')
title3 = ax3.set_title('Presión adimensional')
ax3.set_xlabel('x*')
ax3.set_ylabel('y*')

cont4 = ax4.contourf(X_star, Y_star, tau_history[0], levels=20, cmap='coolwarm')
cbar4 = fig.colorbar(cont4, ax=ax4)
cbar4.set_label('τ*')
title4 = ax4.set_title('Shear stress adimensional')
ax4.set_xlabel('x*')
ax4.set_ylabel('y*')

plt.tight_layout()

def init():
    return []

def update(frame):
    # Actualizar velocidad
    speed = np.sqrt(u_history[frame]**2 + v_history[frame]**2)
    for coll in ax1.collections:
        coll.remove()
    cont1 = ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
    
    # Actualizar temperatura
    for coll in ax2.collections:
        coll.remove()
    cont2 = ax2.contourf(X_star, Y_star, T_history[frame], levels=20, cmap='jet')
    
    # Actualizar presión
    for coll in ax3.collections:
        coll.remove()
    cont3 = ax3.contourf(X_star, Y_star, p_history[frame], levels=20, cmap='viridis')
    
    # Actualizar shear stress
    for coll in ax4.collections:
        coll.remove()
    cont4 = ax4.contourf(X_star, Y_star, tau_history[frame], levels=20, cmap='coolwarm')
    
    # Actualizar títulos
    t = frame*10*dt_star
    title1.set_text(f'Velocidad (t*={t:.2f})')
    title2.set_text(f'Temperatura (t*={t:.2f})')
    title3.set_text(f'Presión (t*={t:.2f})')
    title4.set_text(f'Shear stress (t*={t:.2f})')
    
    # Retornar todos los artistas actualizados
    return ax1.collections + ax2.collections + ax3.collections + ax4.collections

anim = FuncAnimation(fig, update, frames=len(u_history), init_func=init, interval=100, blit=False)
plt.close()

HTML(anim.to_jshtml())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# =============================================================================
# PARÁMETROS Y CONSTANTES
# =============================================================================
# Propiedades del aire (25°C, 1 atm)
rho = 1.184       # Densidad [kg/m³]
mu = 1.849e-5     # Viscosidad dinámica [Pa·s]
nu = mu/rho       # Viscosidad cinemática [m²/s]
k = 0.0262        # Conductividad térmica [W/(m·K)]
alpha = 2.22e-5   # Difusividad térmica [m²/s]
cp = 1006.0       # Calor específico [J/(kg·K)]

# Escalas características
L = 1.0           # Longitud característica [m]
H = 0.1           # Altura característica [m]
u0 = 2.0          # Velocidad característica [m/s]
up = -1.0         # Velocidad placa [m/s]
T0 = 293.0        # Temperatura referencia [K]
p0 = 101325.0     # Presión referencia [Pa]

# Cálculo de u_c, p_f y T_f
uc = u0 - up      # Velocidad característica corregida
pf = mu*uc*L/H**2 + p0
Tf = T0 + (mu*uc**2)/k

# Parámetros adimensionales
Re = 20.0                     # Reynolds
Pr = 10.0                     # Prandtl
Ec = 0.1                      # Eckert
Eu = 1.0                      # Euler (presión)

# =============================================
# PARÁMETROS ADIMENSIONALES
# =============================================
Lx_star, Ly_star = 1.0, 1.0  # Dimensiones adimensionales
nx, ny = 50, 50               # Puntos en la malla
dx_star = Lx_star/(nx-1)      # Paso espacial x
dy_star = Ly_star/(ny-1)      # Paso espacial y

# Condiciones de frontera
up = -0.5               # Velocidad placa
u0 = 1.0                # Velocidad flujo libre
T0_star = 1.0           # Temperatura pared inferior
T1_star = 0.0           # Temperatura pared superior

# Tiempo
dt_star = 0.001         # Paso temporal
nt = 500                # Pasos totales

# =============================================
# INICIALIZACIÓN
# =============================================
x_star = np.linspace(0, Lx_star, nx)
y_star = np.linspace(0, Ly_star, ny)
X_star, Y_star = np.meshgrid(x_star, y_star)

# Campos iniciales
u_star = np.zeros((ny, nx)) + u0
v_star = np.zeros((ny, nx))
p_star = np.zeros((ny, nx))
T_star = np.zeros((ny, nx)) + T1_star

# Condiciones de frontera
u_star[0, :] = up       # Placa inferior móvil
u_star[-1, :] = u0      # Flujo libre superior
v_star[0, :] = 0        # No penetración inferior
v_star[-1, :] = 0       # No penetración superior
T_star[0, :] = T0_star  # Temperatura pared inferior
T_star[-1, :] = T1_star # Temperatura pared superior

# =============================================
# SIMULACIÓN
# =============================================
u_history, v_history, p_history, T_history, tau_history = [], [], [], [], []

for n in range(nt):
    u_old, v_old, p_old, T_old = u_star.copy(), v_star.copy(), p_star.copy(), T_star.copy()
    
    # Primero resolvemos las ecuaciones de momento
    for i in range(1, ny-1):
        for j in range(1, nx-1):
            # Términos convectivos
            conv_u_x = u_old[i,j] * (u_old[i,j+1] - u_old[i,j-1])/(2*dx_star)
            conv_u_y = v_old[i,j] * (u_old[i+1,j] - u_old[i-1,j])/(2*dy_star)
            
            # Términos difusivos
            diff_u_x = (u_old[i,j+1] - 2*u_old[i,j] + u_old[i,j-1])/(dx_star**2)
            diff_u_y = (u_old[i+1,j] - 2*u_old[i,j] + u_old[i-1,j])/(dy_star**2)
            
            # Gradiente de presión
            # grad_p_x = (p_old[i,j+1] - p_old[i,j-1])/(2*dx_star)
            grad_p_x = 0
            
            # Ecuación de momento x (actualizada)
            u_star[i,j] = u_old[i,j] + dt_star*(-conv_u_x - conv_u_y - Eu*grad_p_x + (1/Re)*(diff_u_x + diff_u_y))
            
            # Ecuación de momento y
            conv_v_x = u_old[i,j] * (v_old[i,j+1] - v_old[i,j-1])/(2*dx_star)
            conv_v_y = v_old[i,j] * (v_old[i+1,j] - v_old[i-1,j])/(2*dy_star)
            
            diff_v_x = (v_old[i,j+1] - 2*v_old[i,j] + v_old[i,j-1])/(dx_star**2)
            diff_v_y = (v_old[i+1,j] - 2*v_old[i,j] + v_old[i-1,j])/(dy_star**2)
            
            grad_p_y = (p_old[i+1,j] - p_old[i-1,j])/(2*dy_star)
            
            v_star[i,j] = v_old[i,j] + dt_star*(-conv_v_x - conv_v_y - Eu*grad_p_y + (1/Re)*(diff_v_x + diff_v_y))
    
    # Resolver ecuación de continuidad para presión (método de proyección)
    for _ in range(20):  # Iteraciones para presión
        for i in range(1, ny-1):
            for j in range(1, nx-1):
                p_star[i,j] = 0.25*(p_old[i+1,j] + p_old[i-1,j] + p_old[i,j+1] + p_old[i,j-1] - 
                              (dx_star*dy_star)/(2*(dx_star**2 + dy_star**2)) * (
                                  (u_star[i,j+1] - u_star[i,j-1])/(2*dx_star) + 
                                  (v_star[i+1,j] - v_star[i-1,j])/(2*dy_star)))
    
    # Actualizar velocidades con nuevo gradiente de presión
    for i in range(1, ny-1):
        for j in range(1, nx-1):
            u_star[i,j] -= dt_star * Eu * (p_star[i,j+1] - p_star[i,j-1])/(2*dx_star)
            v_star[i,j] -= dt_star * Eu * (p_star[i+1,j] - p_star[i-1,j])/(2*dy_star)
    
    # Resolver ecuación de energía
    for i in range(1, ny-1):
        for j in range(1, nx-1):
            # Términos convectivos
            conv_T_x = u_star[i,j] * (T_old[i,j+1] - T_old[i,j-1])/(2*dx_star)
            conv_T_y = v_star[i,j] * (T_old[i+1,j] - T_old[i-1,j])/(2*dy_star)
            
            # Términos difusivos
            diff_T_x = (T_old[i,j+1] - 2*T_old[i,j] + T_old[i,j-1])/(dx_star**2)
            diff_T_y = (T_old[i+1,j] - 2*T_old[i,j] + T_old[i-1,j])/(dy_star**2)
            
            # Disipación viscosa
            Sxx = (u_star[i,j+1] - u_star[i,j-1])/(2*dx_star)
            Syy = (v_star[i+1,j] - v_star[i-1,j])/(2*dy_star)
            Sxy = 0.5*((u_star[i+1,j] - u_star[i-1,j])/(2*dy_star) + 
                       (v_star[i,j+1] - v_star[i,j-1])/(2*dx_star))
            
            viscous_heating = (Ec/Re)*(2*(Sxx**2 + Syy**2) + 4*Sxy**2)
            
            # Ecuación de energía
            T_star[i,j] = T_old[i,j] + dt_star*(-conv_T_x - conv_T_y + 
                                                (1/(Re*Pr))*(diff_T_x + diff_T_y) + 
                                                viscous_heating)
    
    # Calcular shear stress en todo el dominio (τ = μ(du/dy))
    tau = np.zeros((ny, nx))
    for i in range(1, ny-1):
        for j in range(nx):
            tau[i,j] = (u_star[i+1,j] - u_star[i-1,j])/(2*dy_star)  # du/dy
    
    # Condiciones de frontera en paredes laterales (Neumann en la pared derecha)
    v_star[:, -1] = 0  # No penetración en la pared lateral derecha
    u_star[-1, :] = u0  # Flujo libre superior
    u_star[0, :] = up   # Placa inferior móvil
    v_star[0, :] = 0    # No penetración inferior
    v_star[-1, :] = 0   # No penetración superior

    # Cambiar la condición de frontera en la pared lateral derecha para Neumann
    u_star[:, -1] = u_star[:, -2]  # ∂u/∂x = 0 en la pared derecha (Neumann)

    
    # Condiciones de frontera para presión (Neumann)
    p_star[:, 0] = p_star[:, 1]    # ∂p/∂x = 0 en x=0
    p_star[:, -1] = p_star[:, -2]  # ∂p/∂x = 0 en x=Lx
    p_star[0, :] = p_star[1, :]    # ∂p/∂y = 0 en y=0
    p_star[-1, :] = p_star[-2, :]  # ∂p/∂y = 0 en y=Ly
    
    if n % 10 == 0:
        u_history.append(u_star.copy())
        v_history.append(v_star.copy())
        p_history.append(p_star.copy())
        T_history.append(T_star.copy())
        tau_history.append(tau.copy())

# =============================================
# VISUALIZACIÓN ANIMADA
# =============================================
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))

# Configuración inicial
speed = np.sqrt(u_history[0]**2 + v_history[0]**2)
cont1 = ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
cbar1 = fig.colorbar(cont1, ax=ax1)
cbar1.set_label('|u*|')
title1 = ax1.set_title(f'Velocidad (t*=0.00)')
ax1.set_xlabel('x*')
ax1.set_ylabel('y*')

cont2 = ax2.contourf(X_star, Y_star, T_history[0], levels=20, cmap='jet')
cbar2 = fig.colorbar(cont2, ax=ax2)
cbar2.set_label('T*')
title2 = ax2.set_title('Temperatura adimensional')
ax2.set_xlabel('x*')
ax2.set_ylabel('y*')

cont3 = ax3.contourf(X_star, Y_star, p_history[0], levels=20, cmap='viridis')
cbar3 = fig.colorbar(cont3, ax=ax3)
cbar3.set_label('p*')
title3 = ax3.set_title('Presión adimensional')
ax3.set_xlabel('x*')
ax3.set_ylabel('y*')

cont4 = ax4.contourf(X_star, Y_star, tau_history[0], levels=20, cmap='coolwarm')
cbar4 = fig.colorbar(cont4, ax=ax4)
cbar4.set_label('τ*')
title4 = ax4.set_title('Shear stress adimensional')
ax4.set_xlabel('x*')
ax4.set_ylabel('y*')

plt.tight_layout()

def init():
    return []

def update(frame):
    # Actualizar velocidad
    speed = np.sqrt(u_history[frame]**2 + v_history[frame]**2)
    for coll in ax1.collections:
        coll.remove()
    cont1 = ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
    
    # Actualizar temperatura
    for coll in ax2.collections:
        coll.remove()
    cont2 = ax2.contourf(X_star, Y_star, T_history[frame], levels=20, cmap='jet')
    
    # Actualizar presión
    for coll in ax3.collections:
        coll.remove()
    cont3 = ax3.contourf(X_star, Y_star, p_history[frame], levels=20, cmap='viridis')
    
    # Actualizar shear stress
    for coll in ax4.collections:
        coll.remove()
    cont4 = ax4.contourf(X_star, Y_star, tau_history[frame], levels=20, cmap='coolwarm')
    
    # Actualizar títulos
    t = frame*10*dt_star
    title1.set_text(f'Velocidad (t*={t:.2f})')
    title2.set_text(f'Temperatura (t*={t:.2f})')
    title3.set_text(f'Presión (t*={t:.2f})')
    title4.set_text(f'Shear stress (t*={t:.2f})')
    
    # Retornar todos los artistas actualizados
    return ax1.collections + ax2.collections + ax3.collections + ax4.collections

anim = FuncAnimation(fig, update, frames=len(u_history), init_func=init, interval=100, blit=False)
plt.close()

HTML(anim.to_jshtml())
```
