```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from IPython.display import display, HTML
import time
import pickle
import os
import sys
import csv
```

```python
def run_simulation(nx, ny):

    # --- Parámetros adimensionales ---
    Re, Pr, Ec, Eu = 20.0, 10.0, 0.1, 1.0
    Lx_star, Ly_star = 1.0, 1.0
    dx_star = Lx_star / (nx - 1)
    dy_star = Ly_star / (ny - 1)
    up, u0 = -0.5, 1.0
    T0_star, T1_star = 0.0, 0.0

    # --- CFL-SAFE TIME STEP (STATIC) ---
    cfl_limit = 0.1  # Muy conservador para evitar overflows en altas resoluciones
    u_max_expected = 2.0
    v_max_expected = 2.0
    dt_star = cfl_limit * min(dx_star / u_max_expected, dy_star / v_max_expected)

    # Límite superior para dt_star en simulaciones muy finas
    dt_star = min(dt_star, 1e-3)

    # Garantizar mínimo 1100 pasos (para llegar a t* = 1)
    nt = max(1100, int(1.0 / dt_star))

    # --- Malla y campos ---
    x_star = np.linspace(0, Lx_star, nx)
    y_star = np.linspace(0, Ly_star, ny)
    X_star, Y_star = np.meshgrid(x_star, y_star)

    u_star = np.zeros((ny, nx), dtype=np.float64) + u0
    v_star = np.zeros((ny, nx), dtype=np.float64)
    p_star = np.zeros((ny, nx), dtype=np.float64)
    T_star = np.zeros((ny, nx), dtype=np.float64) + T1_star

    # --- Condiciones de frontera iniciales ---
    u_star[0, :] = up
    u_star[-1, :] = u0
    v_star[0, :] = v_star[-1, :] = 0
    T_star[0, :] = T0_star
    T_star[-1, :] = T1_star

    u_history, v_history, p_history, T_history, tau_history = [], [], [], [], []

    for n in range(nt):
        u_old, v_old, p_old, T_old = u_star.copy(), v_star.copy(), p_star.copy(), T_star.copy()

        # --- Ecuaciones de momento ---
        for i in range(1, ny - 1):
            for j in range(1, nx - 1):
                conv_u_x = u_old[i, j] * (u_old[i, j + 1] - u_old[i, j - 1]) / (2 * dx_star)
                conv_u_y = v_old[i, j] * (u_old[i + 1, j] - u_old[i - 1, j]) / (2 * dy_star)
                diff_u = ((u_old[i, j + 1] - 2 * u_old[i, j] + u_old[i, j - 1]) / dx_star**2 +
                          (u_old[i + 1, j] - 2 * u_old[i, j] + u_old[i - 1, j]) / dy_star**2)
                u_star[i, j] = u_old[i, j] + dt_star * (-conv_u_x - conv_u_y + (1 / Re) * diff_u)

                conv_v_x = u_old[i, j] * (v_old[i, j + 1] - v_old[i, j - 1]) / (2 * dx_star)
                conv_v_y = v_old[i, j] * (v_old[i + 1, j] - v_old[i - 1, j]) / (2 * dy_star)
                diff_v = ((v_old[i, j + 1] - 2 * v_old[i, j] + v_old[i, j - 1]) / dx_star**2 +
                          (v_old[i + 1, j] - 2 * v_old[i, j] + v_old[i - 1, j]) / dy_star**2)
                grad_p_y = (p_old[i + 1, j] - p_old[i - 1, j]) / (2 * dy_star)
                v_star[i, j] = v_old[i, j] + dt_star * (-conv_v_x - conv_v_y - Eu * grad_p_y + (1 / Re) * diff_v)

        # --- Presión ---
        for _ in range(200):
            for i in range(1, ny - 1):
                for j in range(1, nx - 1):
                    rhs = ((u_star[i, j + 1] - u_star[i, j - 1]) / (2 * dx_star) +
                           (v_star[i + 1, j] - v_star[i - 1, j]) / (2 * dy_star))
                    p_star[i, j] = 0.25 * (p_old[i + 1, j] + p_old[i - 1, j] + p_old[i, j + 1] + p_old[i, j - 1] -
                                          (dx_star * dy_star) / (2 * (dx_star**2 + dy_star**2)) * rhs)

        # --- Corrección de velocidades ---
        for i in range(1, ny - 1):
            for j in range(1, nx - 1):
                u_star[i, j] -= dt_star * Eu * (p_star[i, j + 1] - p_star[i, j - 1]) / (2 * dx_star)
                v_star[i, j] -= dt_star * Eu * (p_star[i + 1, j] - p_star[i - 1, j]) / (2 * dy_star)

        # --- Ecuación de energía ---
        for i in range(1, ny - 1):
            for j in range(1, nx - 1):
                conv_T_x = u_star[i, j] * (T_old[i, j + 1] - T_old[i, j - 1]) / (2 * dx_star)
                conv_T_y = v_star[i, j] * (T_old[i + 1, j] - T_old[i - 1, j]) / (2 * dy_star)
                diff_T = ((T_old[i, j + 1] - 2 * T_old[i, j] + T_old[i, j - 1]) / dx_star**2 +
                          (T_old[i + 1, j] - 2 * T_old[i, j] + T_old[i - 1, j]) / dy_star**2)
                Sxx = (u_star[i, j + 1] - u_star[i, j - 1]) / (2 * dx_star)
                Syy = (v_star[i + 1, j] - v_star[i - 1, j]) / (2 * dy_star)
                Sxy = 0.5 * ((u_star[i + 1, j] - u_star[i - 1, j]) / (2 * dy_star) +
                             (v_star[i, j + 1] - v_star[i, j - 1]) / (2 * dx_star))
                viscous_heating = (Ec / Re) * (2 * (Sxx**2 + Syy**2) + 4 * Sxy**2)
                T_star[i, j] = T_old[i, j] + dt_star * (-conv_T_x - conv_T_y + (1 / (Re * Pr)) * diff_T + viscous_heating)

        # --- Shear stress ---
        tau = np.zeros_like(u_star)
        tau[1:-1, :] = (u_star[2:, :] - u_star[:-2, :]) / (2 * dy_star)

        # --- Condiciones de frontera ---
        u_star[0, :] = up
        u_star[-1, :] = u0
        u_star[:, -1] = u_star[:, -2]
        v_star[0, :] = v_star[-1, :] = 0
        v_star[:, -1] = 0
        p_star[:, 0] = p_star[:, 1]
        p_star[:, -1] = p_star[:, -2]
        p_star[0, :] = p_star[1, :]
        p_star[-1, :] = p_star[-2, :]

        # --- Historial ---
        if n % 10 == 0:
            print(f"\rPaso {n}/{nt} (dt*={dt_star:.5f})", end="")
            sys.stdout.flush()
            u_history.append(u_star.copy())
            v_history.append(v_star.copy())
            p_history.append(p_star.copy())
            T_history.append(T_star.copy())
            tau_history.append(tau.copy())

    print("\rSimulación finalizada.                      ")

    return {
        "u_history": u_history,
        "v_history": v_history,
        "p_history": p_history,
        "T_history": T_history,
        "tau_history": tau_history,
        "params": {"nx": nx, "ny": ny, "dt_star": dt_star, "nt": nt},
        "X_star": X_star,
        "Y_star": Y_star
    }
```

```python
nx = ny = 25

result = run_simulation(nx=nx, ny=ny)

u_history = result["u_history"]
v_history = result["v_history"]
p_history = result["p_history"]
T_history = result["T_history"]
tau_history = result["tau_history"]
X_star = result["X_star"]
Y_star = result["Y_star"]
dt_star = result["params"]["dt_star"]


# =============================================
# VISUALIZACIÓN ANIMADA (t* llega hasta 1.0)
# =============================================

# Paso visual equivalente (cada 10 pasos)
dt_star_visual = 10 * dt_star

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))

# Configuración inicial
speed = np.sqrt(u_history[0]**2 + v_history[0]**2)
cont1 = ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
cbar1 = fig.colorbar(cont1, ax=ax1)
cbar1.set_label('|u*|')
title1 = ax1.set_title(f'Velocidad (t*=0.00)')
ax1.set_xlabel('x*')
ax1.set_ylabel('y*')

cont2 = ax2.contourf(X_star, Y_star, T_history[0], levels=20, cmap='jet')
cbar2 = fig.colorbar(cont2, ax=ax2)
cbar2.set_label('T*')
title2 = ax2.set_title('Temperatura adimensional')
ax2.set_xlabel('x*')
ax2.set_ylabel('y*')

cont3 = ax3.contourf(X_star, Y_star, p_history[0], levels=20, cmap='viridis')
cbar3 = fig.colorbar(cont3, ax=ax3)
cbar3.set_label('p*')
title3 = ax3.set_title('Presión adimensional')
ax3.set_xlabel('x*')
ax3.set_ylabel('y*')

cont4 = ax4.contourf(X_star, Y_star, tau_history[0], levels=20, cmap='coolwarm')
cbar4 = fig.colorbar(cont4, ax=ax4)
cbar4.set_label('τ*')
title4 = ax4.set_title('Shear stress adimensional')
ax4.set_xlabel('x*')
ax4.set_ylabel('y*')
fig.suptitle(f"Simulación con resolución espacial {nx}x{ny}, dt* = {dt_star:.5f}", fontsize=14)


plt.tight_layout()

def update(frame):
    speed = np.sqrt(u_history[frame]**2 + v_history[frame]**2)

    for ax in [ax1, ax2, ax3, ax4]:
        for coll in ax.collections:
            coll.remove()

    ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
    ax2.contourf(X_star, Y_star, T_history[frame], levels=20, cmap='jet')
    ax3.contourf(X_star, Y_star, p_history[frame], levels=20, cmap='viridis')
    ax4.contourf(X_star, Y_star, tau_history[frame], levels=20, cmap='coolwarm')

    t_star = frame * dt_star_visual
    title1.set_text(f'Velocidad (t*={t_star:.2f})')
    title2.set_text(f'Temperatura (t*={t_star:.2f})')
    title3.set_text(f'Presión (t*={t_star:.2f})')
    title4.set_text(f'Shear stress (t*={t_star:.2f})')

    return ax1.collections + ax2.collections + ax3.collections + ax4.collections

anim = FuncAnimation(fig, update, frames=len(u_history), interval=100, blit=False)
plt.close()

HTML(anim.to_jshtml())
```

```python
def simular_y_animar(resoluciones, output_dir="animaciones"):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    for (nx, ny) in resoluciones:
        print(f"\n🔄 Ejecutando simulación para resolución {nx}x{ny}...")
        result = run_simulation(nx=nx, ny=ny)

        u_history = result["u_history"]
        v_history = result["v_history"]
        p_history = result["p_history"]
        T_history = result["T_history"]
        tau_history = result["tau_history"]
        X_star = result["X_star"]
        Y_star = result["Y_star"]

        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))
        fig.suptitle(f"Simulación con resolución {nx}x{ny}", fontsize=14)

        speed = np.sqrt(u_history[0]**2 + v_history[0]**2)
        cont1 = ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
        cont2 = ax2.contourf(X_star, Y_star, T_history[0], levels=20, cmap='jet')
        cont3 = ax3.contourf(X_star, Y_star, p_history[0], levels=20, cmap='viridis')
        cont4 = ax4.contourf(X_star, Y_star, tau_history[0], levels=20, cmap='coolwarm')

        fig.colorbar(cont1, ax=ax1).set_label('|u*|')
        fig.colorbar(cont2, ax=ax2).set_label('T*')
        fig.colorbar(cont3, ax=ax3).set_label('p*')
        fig.colorbar(cont4, ax=ax4).set_label('τ*')

        x_ticks = np.linspace(0, 1, nx)
        y_ticks = np.linspace(0, 1, ny)
        x_tick_freq = max(nx // 10, 1)
        y_tick_freq = max(ny // 10, 1)
        x_tick_labels = [f"{x:.2f}" if i % x_tick_freq == 0 else "" for i, x in enumerate(x_ticks)]
        y_tick_labels = [f"{y:.2f}" if i % y_tick_freq == 0 else "" for i, y in enumerate(y_ticks)]

        def update(frame):
            speed = np.sqrt(u_history[frame]**2 + v_history[frame]**2)
            t_star = frame / (len(u_history) - 1)

            for ax in [ax1, ax2, ax3, ax4]:
                ax.clear()
                ax.set_xticks(x_ticks)
                ax.set_yticks(y_ticks)
                ax.set_xticklabels(x_tick_labels, fontsize=7)
                ax.set_yticklabels(y_tick_labels, fontsize=7)
                ax.grid(True, which='both', linestyle='--', linewidth=0.5)

            ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
            ax2.contourf(X_star, Y_star, T_history[frame], levels=20, cmap='jet')
            ax3.contourf(X_star, Y_star, p_history[frame], levels=20, cmap='viridis')
            ax4.contourf(X_star, Y_star, tau_history[frame], levels=20, cmap='coolwarm')

            ax1.set_title(f'|u*| (t*={t_star:.2f})')
            ax2.set_title(f'T* (t*={t_star:.2f})')
            ax3.set_title(f'p* (t*={t_star:.2f})')
            ax4.set_title(f'τ* (t*={t_star:.2f})')

            for ax in [ax1, ax2, ax3, ax4]:
                ax.set_xlabel("x*")
                ax.set_ylabel("y*")

            return []

        anim = FuncAnimation(fig, update, frames=len(u_history), interval=100, blit=False)

        filename = f"simulacion_{nx}x{ny}.mp4"
        path = os.path.join(output_dir, filename)

        print(f"💾 Guardando animación en {path}")
        writer = FFMpegWriter(fps=10)
        anim.save(path, writer=writer)
        plt.close()
```

```python
def simular_y_animar_con_guardado(resoluciones, output_dir="animaciones", grid_resolution=None):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    for (nx, ny) in resoluciones:
        print(f"\n🔄 Ejecutando simulación para resolución {nx}x{ny}...")
        result = run_simulation(nx=nx, ny=ny)

        u_history = result["u_history"]
        v_history = result["v_history"]
        p_history = result["p_history"]
        T_history = result["T_history"]
        tau_history = result["tau_history"]
        X_star = result["X_star"]
        Y_star = result["Y_star"]
        dt_star = result["params"]["dt_star"]

        # Guardar dt_star
        dt_file = os.path.join(output_dir, f"dt_star_{nx}x{ny}.txt")
        with open(dt_file, "w") as f:
            f.write(f"{dt_star:.10f}")

        # Guardar resultados en CSV
        for name, history in zip(["u", "v", "p", "T", "tau"],
                                 [u_history, v_history, p_history, T_history, tau_history]):
            for t, frame in enumerate(history):
                filename = f"{name}_{nx}x{ny}_t{t:04d}.csv"
                filepath = os.path.join(output_dir, filename)
                np.savetxt(filepath, frame, delimiter=",", fmt="%.6e")

        # Crear animación
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))
        fig.suptitle(f"Resolución {nx}x{ny}, dt* = {dt_star:.5e}", fontsize=14)

        speed = np.sqrt(u_history[0]**2 + v_history[0]**2)
        cont1 = ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
        cont2 = ax2.contourf(X_star, Y_star, T_history[0], levels=20, cmap='jet')
        cont3 = ax3.contourf(X_star, Y_star, p_history[0], levels=20, cmap='viridis')
        cont4 = ax4.contourf(X_star, Y_star, tau_history[0], levels=20, cmap='coolwarm')

        fig.colorbar(cont1, ax=ax1).set_label('|u*|')
        fig.colorbar(cont2, ax=ax2).set_label('T*')
        fig.colorbar(cont3, ax=ax3).set_label('p*')
        fig.colorbar(cont4, ax=ax4).set_label('τ*')

        # Etiquetas
        x_ticks = np.linspace(0, 1, grid_resolution if grid_resolution else nx)
        y_ticks = np.linspace(0, 1, grid_resolution if grid_resolution else ny)
        x_tick_labels = [f"{x:.2f}" for x in x_ticks]
        y_tick_labels = [f"{y:.2f}" for y in y_ticks]

        def update(frame):
            speed = np.sqrt(u_history[frame]**2 + v_history[frame]**2)
            t_star = frame / (len(u_history) - 1)

            for ax in [ax1, ax2, ax3, ax4]:
                ax.clear()
                ax.set_xticks(x_ticks)
                ax.set_yticks(y_ticks)
                ax.set_xticklabels(x_tick_labels, fontsize=7)
                ax.set_yticklabels(y_tick_labels, fontsize=7)
                ax.grid(True, which='both', linestyle='--', linewidth=0.5)

            ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
            ax2.contourf(X_star, Y_star, T_history[frame], levels=20, cmap='jet')
            ax3.contourf(X_star, Y_star, p_history[frame], levels=20, cmap='viridis')
            ax4.contourf(X_star, Y_star, tau_history[frame], levels=20, cmap='coolwarm')

            ax1.set_title(f'|u*| (t*={t_star:.2f})')
            ax2.set_title(f'T* (t*={t_star:.2f})')
            ax3.set_title(f'p* (t*={t_star:.2f})')
            ax4.set_title(f'τ* (t*={t_star:.2f})')

            for ax in [ax1, ax2, ax3, ax4]:
                ax.set_xlabel("x*")
                ax.set_ylabel("y*")

            return []

        anim = FuncAnimation(fig, update, frames=len(u_history), interval=100, blit=False)

        filename = f"simulacion_{nx}x{ny}.mp4"
        path = os.path.join(output_dir, filename)

        print(f"💾 Guardando animación en {path}")
        writer = FFMpegWriter(fps=10)
        anim.save(path, writer=writer)
        plt.close()
```

```python
def simular_y_animar_con_guardado(resoluciones, output_dir="resultados_grid_conve", grid_resolution="match"):
    os.makedirs(output_dir, exist_ok=True)

    for (nx, ny) in resoluciones:
        print(f"\n🔄 Ejecutando simulación para resolución {nx}x{ny}...")
        result = run_simulation(nx=nx, ny=ny)

        carpeta = os.path.join(output_dir, f"{nx}x{ny}")
        os.makedirs(carpeta, exist_ok=True)

        # Guardar CSVs
        def guardar_csv(data_history, nombre):
            with open(os.path.join(carpeta, f"{nombre}.csv"), "w", newline="") as f:
                writer = csv.writer(f)
                for frame in data_history:
                    writer.writerow(frame.flatten())

        guardar_csv(result["u_history"], "u")
        guardar_csv(result["v_history"], "v")
        guardar_csv(result["p_history"], "p")
        guardar_csv(result["T_history"], "T")
        guardar_csv(result["tau_history"], "tau")

        # Guardar dt_star
        with open(os.path.join(carpeta, "dt_star.txt"), "w") as f:
            f.write(f"{result['params']['dt_star']:.10f}")

        # Animación
        u_history = result["u_history"]
        v_history = result["v_history"]
        p_history = result["p_history"]
        T_history = result["T_history"]
        tau_history = result["tau_history"]
        X_star = result["X_star"]
        Y_star = result["Y_star"]
        dt_star = result["params"]["dt_star"]

        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))
        fig.suptitle(f"Resolución {nx}x{ny}, dt* = {dt_star:.5e}", fontsize=14)

        speed = np.sqrt(u_history[0]**2 + v_history[0]**2)
        cont1 = ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
        cont2 = ax2.contourf(X_star, Y_star, T_history[0], levels=20, cmap='jet')
        cont3 = ax3.contourf(X_star, Y_star, p_history[0], levels=20, cmap='viridis')
        cont4 = ax4.contourf(X_star, Y_star, tau_history[0], levels=20, cmap='coolwarm')

        fig.colorbar(cont1, ax=ax1).set_label('|u*|')
        fig.colorbar(cont2, ax=ax2).set_label('T*')
        fig.colorbar(cont3, ax=ax3).set_label('p*')
        fig.colorbar(cont4, ax=ax4).set_label('τ*')

        if grid_resolution == "match":
            x_ticks = np.linspace(0, 1, nx)
            y_ticks = np.linspace(0, 1, ny)
        else:
            x_ticks = np.linspace(0, 1, grid_resolution)
            y_ticks = np.linspace(0, 1, grid_resolution)

        x_tick_labels = [f"{x:.2f}" for x in x_ticks]
        y_tick_labels = [f"{y:.2f}" for y in y_ticks]

        def update(frame):
            speed = np.sqrt(u_history[frame]**2 + v_history[frame]**2)
            t_star = frame / (len(u_history) - 1)

            for ax in [ax1, ax2, ax3, ax4]:
                ax.clear()
                ax.set_xticks(x_ticks)
                ax.set_yticks(y_ticks)
                ax.set_xticklabels(x_tick_labels, fontsize=7)
                ax.set_yticklabels(y_tick_labels, fontsize=7)
                ax.grid(True, linestyle='--', linewidth=0.5)

            ax1.contourf(X_star, Y_star, speed, levels=20, cmap='jet')
            ax2.contourf(X_star, Y_star, T_history[frame], levels=20, cmap='jet')
            ax3.contourf(X_star, Y_star, p_history[frame], levels=20, cmap='viridis')
            ax4.contourf(X_star, Y_star, tau_history[frame], levels=20, cmap='coolwarm')

            ax1.set_title(f'|u*| (t*={t_star:.2f})')
            ax2.set_title(f'T* (t*={t_star:.2f})')
            ax3.set_title(f'p* (t*={t_star:.2f})')
            ax4.set_title(f'τ* (t*={t_star:.2f})')

            for ax in [ax1, ax2, ax3, ax4]:
                ax.set_xlabel("x*")
                ax.set_ylabel("y*")

            return []

        anim = FuncAnimation(fig, update, frames=len(u_history), interval=100, blit=False)

        video_path = os.path.join(carpeta, f"simulacion_{nx}x{ny}.mp4")
        print(f"💾 Guardando animación en {video_path}")
        writer = FFMpegWriter(fps=10)
        anim.save(video_path, writer=writer)
        plt.close()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from IPython.display import display, HTML
import time
import pickle
import os
import sys
import csv
## Funcion simulacion
def run_simulation(nx=50, ny=50):
    # Parámetros adimensionales
    Re, Pr, Ec, Eu = 20.0, 10.0, 0.1, 1.0
    Lx_star, Ly_star = 1.0, 1.0
    dx_star = Lx_star / (nx - 1)
    dy_star = Ly_star / (ny - 1)

    # Cálculo conservador de dt_star por CFL
    CFL_target = 0.0446
    u_max = 1.0
    dt_cfl = CFL_target * dx_star / u_max
    dt_max = 1e-5  # opcional: evita explosiones en mallas gruesas
    dt_star = min(dt_cfl, dt_max)

    # Ajustar dt_star para que nt sea entero y t* = 1.0 exacto
    nt_min = 1100  # por precisión
    nt = max(nt_min, int(np.ceil(1.0 / dt_star)))
    dt_star = 1.0 / nt  # se reajusta para llegar exactamente a t*=1


    up, u0 = -0.5, 1.0
    T0_star, T1_star = 0.0, 0.0

    x_star = np.linspace(0, Lx_star, nx)
    y_star = np.linspace(0, Ly_star, ny)
    X_star, Y_star = np.meshgrid(x_star, y_star)

    u_star = np.zeros((ny, nx)) + u0
    v_star = np.zeros((ny, nx))
    p_star = np.zeros((ny, nx))
    T_star = np.zeros((ny, nx)) + T1_star

    # Condiciones de frontera
    u_star[0, :] = up
    u_star[-1, :] = u0
    v_star[0, :] = v_star[-1, :] = 0
    T_star[0, :] = T0_star
    T_star[-1, :] = T1_star

    u_history, v_history, p_history, T_history, tau_history = [], [], [], [], []

    for n in range(nt):
        u_old, v_old, p_old, T_old = u_star.copy(), v_star.copy(), p_star.copy(), T_star.copy()
        
        for i in range(1, ny-1):
            for j in range(1, nx-1):
                conv_u_x = u_old[i,j] * (u_old[i,j+1] - u_old[i,j-1]) / (2*dx_star)
                conv_u_y = v_old[i,j] * (u_old[i+1,j] - u_old[i-1,j]) / (2*dy_star)
                diff_u_x = (u_old[i,j+1] - 2*u_old[i,j] + u_old[i,j-1]) / dx_star**2
                diff_u_y = (u_old[i+1,j] - 2*u_old[i,j] + u_old[i-1,j]) / dy_star**2
                grad_p_x = 0  # omitido
                u_star[i,j] += dt_star * (-conv_u_x - conv_u_y - Eu * grad_p_x + (1/Re)*(diff_u_x + diff_u_y))

                conv_v_x = u_old[i,j] * (v_old[i,j+1] - v_old[i,j-1]) / (2*dx_star)
                conv_v_y = v_old[i,j] * (v_old[i+1,j] - v_old[i-1,j]) / (2*dy_star)
                diff_v_x = (v_old[i,j+1] - 2*v_old[i,j] + v_old[i,j-1]) / dx_star**2
                diff_v_y = (v_old[i+1,j] - 2*v_old[i,j] + v_old[i-1,j]) / dy_star**2
                grad_p_y = (p_old[i+1,j] - p_old[i-1,j]) / (2*dy_star)
                v_star[i,j] += dt_star * (-conv_v_x - conv_v_y - Eu * grad_p_y + (1/Re)*(diff_v_x + diff_v_y))

        for _ in range(20):
            for i in range(1, ny-1):
                for j in range(1, nx-1):
                    p_star[i,j] = 0.25 * (p_old[i+1,j] + p_old[i-1,j] + p_old[i,j+1] + p_old[i,j-1] - 
                                          (dx_star*dy_star)/(2*(dx_star**2 + dy_star**2)) * (
                                              (u_star[i,j+1] - u_star[i,j-1])/(2*dx_star) + 
                                              (v_star[i+1,j] - v_star[i-1,j])/(2*dy_star)))

        for i in range(1, ny-1):
            for j in range(1, nx-1):
                u_star[i,j] -= dt_star * Eu * (p_star[i,j+1] - p_star[i,j-1])/(2*dx_star)
                v_star[i,j] -= dt_star * Eu * (p_star[i+1,j] - p_star[i-1,j])/(2*dy_star)

        for i in range(1, ny-1):
            for j in range(1, nx-1):
                conv_T_x = u_star[i,j] * (T_old[i,j+1] - T_old[i,j-1])/(2*dx_star)
                conv_T_y = v_star[i,j] * (T_old[i+1,j] - T_old[i-1,j])/(2*dy_star)
                diff_T_x = (T_old[i,j+1] - 2*T_old[i,j] + T_old[i,j-1])/(dx_star**2)
                diff_T_y = (T_old[i+1,j] - 2*T_old[i,j] + T_old[i-1,j])/(dy_star**2)
                Sxx = (u_star[i,j+1] - u_star[i,j-1])/(2*dx_star)
                Syy = (v_star[i+1,j] - v_star[i-1,j])/(2*dy_star)
                Sxy = 0.5*((u_star[i+1,j] - u_star[i-1,j])/(2*dy_star) + 
                           (v_star[i,j+1] - v_star[i,j-1])/(2*dx_star))
                viscous_heating = (Ec/Re)*(2*(Sxx**2 + Syy**2) + 4*Sxy**2)
                T_star[i,j] += dt_star * (-conv_T_x - conv_T_y + 
                                          (1/(Re*Pr))*(diff_T_x + diff_T_y) + 
                                          viscous_heating)

        tau = np.zeros((ny, nx))
        for i in range(1, ny-1):
            for j in range(nx):
                tau[i,j] = (u_star[i+1,j] - u_star[i-1,j])/(2*dy_star)

        # Condiciones de frontera
        v_star[:, -1] = 0
        u_star[-1, :] = u0
        u_star[0, :] = up
        v_star[0, :] = v_star[-1, :] = 0
        u_star[:, -1] = u_star[:, -2]
        p_star[:, 0] = p_star[:, 1]
        p_star[:, -1] = p_star[:, -2]
        p_star[0, :] = p_star[1, :]
        p_star[-1, :] = p_star[-2, :]
        if n % 10 == 0:
                    print(f"\rSimulación en progreso: paso {n}/{nt}", end="")
                    sys.stdout.flush()
                    u_history.append(u_star.copy())
                    v_history.append(v_star.copy())
                    p_history.append(p_star.copy())
                    T_history.append(T_star.copy())
                    tau_history.append(tau.copy())
    print("\rSimulación finalizada.                      ")

    return {
        "u_history": u_history,
        "v_history": v_history,
        "p_history": p_history,
        "T_history": T_history,
        "tau_history": tau_history,
        "params": {"nx": nx, "ny": ny, "dt_star": dt_star, "nt": nt},
        "X_star": X_star,
        "Y_star": Y_star
    }
## Funcion guardar datos simulacion
def guardar_datos_simulacion(resultados_dict, nx, ny, carpeta='simulaciones'):
    if not os.path.exists(carpeta):
        os.makedirs(carpeta)
    nombre_archivo = f"{nx}x{ny}.pkl"
    ruta = os.path.join(carpeta, nombre_archivo)
    with open(ruta, 'wb') as f:
        pickle.dump(resultados_dict, f)
    print(f"✅ Simulación guardada en '{ruta}'")
```

```python
def run_simulation(nx=50, ny=50):

    # Parámetros adimensionales
    Re, Pr, Ec, Eu = 20.0, 10.0, 0.1, 1.0
    Lx_star, Ly_star = 1.0, 1.0
    dx_star = Lx_star / (nx - 1)
    dy_star = Ly_star / (ny - 1)

    # Cálculo conservador de dt_star por CFL
    CFL_target = 0.0446
    u_max = 1.0
    dt_cfl = CFL_target * dx_star / u_max

    # Cuarta sugerencia: ajustar dt_max si la malla es fina
    if nx >= 800:
        dt_max = 5e-7
    else:
        dt_max = 1e-5

    dt_star = min(dt_cfl, dt_max)

    # Ajustar dt_star para que nt sea entero y t* = 1.0 exacto
    nt_min = 1100
    nt = max(nt_min, int(np.ceil(1.0 / dt_star)))
    dt_star = 1.0 / nt

    # Condiciones iniciales y malla
    up, u0 = -0.5, 1.0
    T0_star, T1_star = 0.0, 0.0

    x_star = np.linspace(0, Lx_star, nx)
    y_star = np.linspace(0, Ly_star, ny)
    X_star, Y_star = np.meshgrid(x_star, y_star)

    u_star = np.zeros((ny, nx)) + u0
    v_star = np.zeros((ny, nx))
    p_star = np.zeros((ny, nx))
    T_star = np.zeros((ny, nx)) + T1_star

    # Condiciones de frontera iniciales
    u_star[0, :] = up
    u_star[-1, :] = u0
    v_star[0, :] = v_star[-1, :] = 0
    T_star[0, :] = T0_star
    T_star[-1, :] = T1_star

    u_history, v_history, p_history, T_history, tau_history = [], [], [], [], []

    for n in range(nt):
        u_old, v_old, p_old, T_old = u_star.copy(), v_star.copy(), p_star.copy(), T_star.copy()
        
        for i in range(1, ny - 1):
            for j in range(1, nx - 1):
                conv_u_x = u_old[i, j] * (u_old[i, j + 1] - u_old[i, j - 1]) / (2 * dx_star)
                conv_u_y = v_old[i, j] * (u_old[i + 1, j] - u_old[i - 1, j]) / (2 * dy_star)
                diff_u_x = (u_old[i, j + 1] - 2 * u_old[i, j] + u_old[i, j - 1]) / dx_star**2
                diff_u_y = (u_old[i + 1, j] - 2 * u_old[i, j] + u_old[i - 1, j]) / dy_star**2
                u_star[i, j] += dt_star * (-conv_u_x - conv_u_y + (1 / Re) * (diff_u_x + diff_u_y))

                conv_v_x = u_old[i, j] * (v_old[i, j + 1] - v_old[i, j - 1]) / (2 * dx_star)
                conv_v_y = v_old[i, j] * (v_old[i + 1, j] - v_old[i - 1, j]) / (2 * dy_star)
                diff_v_x = (v_old[i, j + 1] - 2 * v_old[i, j] + v_old[i, j - 1]) / dx_star**2
                diff_v_y = (v_old[i + 1, j] - 2 * v_old[i, j] + v_old[i - 1, j]) / dy_star**2
                grad_p_y = (p_old[i + 1, j] - p_old[i - 1, j]) / (2 * dy_star)
                v_star[i, j] += dt_star * (-conv_v_x - conv_v_y - Eu * grad_p_y + (1 / Re) * (diff_v_x + diff_v_y))

        for _ in range(20):
            for i in range(1, ny - 1):
                for j in range(1, nx - 1):
                    div = (u_star[i, j + 1] - u_star[i, j - 1]) / (2 * dx_star) + (v_star[i + 1, j] - v_star[i - 1, j]) / (2 * dy_star)
                    p_star[i, j] = 0.25 * (p_old[i + 1, j] + p_old[i - 1, j] + p_old[i, j + 1] + p_old[i, j - 1]
                                           - (dx_star * dy_star) / (2 * (dx_star**2 + dy_star**2)) * div)

        for i in range(1, ny - 1):
            for j in range(1, nx - 1):
                u_star[i, j] -= dt_star * Eu * (p_star[i, j + 1] - p_star[i, j - 1]) / (2 * dx_star)
                v_star[i, j] -= dt_star * Eu * (p_star[i + 1, j] - p_star[i - 1, j]) / (2 * dy_star)

        for i in range(1, ny - 1):
            for j in range(1, nx - 1):
                conv_T_x = u_star[i, j] * (T_old[i, j + 1] - T_old[i, j - 1]) / (2 * dx_star)
                conv_T_y = v_star[i, j] * (T_old[i + 1, j] - T_old[i - 1, j]) / (2 * dy_star)
                diff_T_x = (T_old[i, j + 1] - 2 * T_old[i, j] + T_old[i, j - 1]) / dx_star**2
                diff_T_y = (T_old[i + 1, j] - 2 * T_old[i, j] + T_old[i - 1, j]) / dy_star**2
                Sxx = (u_star[i, j + 1] - u_star[i, j - 1]) / (2 * dx_star)
                Syy = (v_star[i + 1, j] - v_star[i - 1, j]) / (2 * dy_star)
                Sxy = 0.5 * ((u_star[i + 1, j] - u_star[i - 1, j]) / (2 * dy_star) + (v_star[i, j + 1] - v_star[i, j - 1]) / (2 * dx_star))
                viscous_heating = (Ec / Re) * (2 * (Sxx**2 + Syy**2) + 4 * Sxy**2)
                T_star[i, j] += dt_star * (-conv_T_x - conv_T_y + (1 / (Re * Pr)) * (diff_T_x + diff_T_y) + viscous_heating)

        tau = np.zeros((ny, nx))
        for i in range(1, ny - 1):
            for j in range(nx):
                tau[i, j] = (u_star[i + 1, j] - u_star[i - 1, j]) / (2 * dy_star)

        # Condiciones de frontera
        v_star[:, -1] = 0
        u_star[-1, :] = u0
        u_star[0, :] = up
        v_star[0, :] = v_star[-1, :] = 0
        u_star[:, -1] = u_star[:, -2]
        p_star[:, 0] = p_star[:, 1]
        p_star[:, -1] = p_star[:, -2]
        p_star[0, :] = p_star[1, :]
        p_star[-1, :] = p_star[-2, :]

        if n % 10 == 0:
            print(f"\rSimulación en progreso: paso {n}/{nt}", end="")
            sys.stdout.flush()
            u_history.append(u_star.copy())
            v_history.append(v_star.copy())
            p_history.append(p_star.copy())
            T_history.append(T_star.copy())
            tau_history.append(tau.copy())

    print("\rSimulación finalizada.                      ")

    return {
        "u_history": u_history,
        "v_history": v_history,
        "p_history": p_history,
        "T_history": T_history,
        "tau_history": tau_history,
        "params": {"nx": nx, "ny": ny, "dt_star": dt_star, "nt": nt},
        "X_star": X_star,
        "Y_star": Y_star
    }
```
