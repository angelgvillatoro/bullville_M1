### **Mathematical Development of the Vorticity Equation for Each Case**

The **vorticity equation** for an incompressible, viscous fluid is derived from the Navier-Stokes equations and is given by:

\[
$\frac{D\vec{\omega}}{Dt} = (\vec{\omega} \cdot \nabla) \vec{u} + \nu \nabla^2 \vec{\omega}$
\]

where:
- \($\frac{D\vec{\omega}}{Dt} = \frac{\partial \vec{\omega}}{\partial t} + (\vec{u} \cdot \nabla) \vec{\omega}$\) (material derivative).
- \($(\vec{\omega} \cdot \nabla) \vec{u}$\) = vortex stretching/tilting term.
- \($\nu \nabla^2 \vec{\omega}$\) = viscous diffusion term.

We analyze both cases in cylindrical coordinates \($(r, \theta, z)$\), suitable for duct flows.

---

## **Case 1: Vortex Ring in a 45° Curved Duct**

### **Assumptions:**
- Incompressible flow (\($\nabla \cdot \vec{u} = 0$\)).
- Axial symmetry (dependence on \($r$\) and \($\theta$\)).
- Duct curvature introduces centrifugal pressure gradients.

### **Mathematical Development:**

1. **Velocity field in the curved duct:**
   - Velocity components: \($u_r$\), \($u_\theta$\), \($u_z$\).
   - Due to curvature, \($u_\theta$\) varies with \($r$\):
     \[
     u_\theta = \Omega r \quad \text{(rotational flow)}
     \]
     where \($\Omega$\) is the angular velocity induced by curvature.

2. **Initial vorticity (axial):**
   \[
   \vec{\omega} = \omega_z \hat{e}_z = (\nabla \times \vec{u})_z \hat{e}_z
   \]

3. **Stretching term \($(\vec{\omega} \cdot \nabla) \vec{u}$\):**
   - In cylindrical coordinates:
     \[
     $(\vec{\omega} \cdot \nabla) \vec{u} = \omega_z \frac{\partial \vec{u}}{\partial z}$
     \]
     - Since the duct is curved, \($\frac{\partial u_\theta}{\partial z} \neq 0$\) (vortex tilting occurs).
     - This leads to **vortex tilting**:
       \[
       $\frac{\partial \omega_\theta}{\partial t} \approx \omega_z \frac{\partial u_\theta}{\partial z}$
       \]

4. **Secondary vorticity generation:**
   - Interaction with curved walls induces **secondary vorticity** (\($\omega_{secondary}$\)) due to boundary layer shear:
     \[
     $\omega_{secondary} \approx \frac{\partial u_\theta}{\partial r} \quad \text{(wall shear effect)}$
     \]

5. **Resulting vorticity equation:**
   \[
   $\frac{D \omega_z}{Dt} = \omega_z \frac{\partial u_z}{\partial z} + \nu \left( \frac{1}{r} \frac{\partial}{\partial r} \left( r \frac{\partial \omega_z}{\partial r} \right) + \frac{\partial^2 \omega_z}{\partial z^2} \right)$
   \]
   - The term \($\omega_z \frac{\partial u_z}{\partial z}$\) represents **axial stretching**.
   - The viscous term \($\nu \nabla^2 \omega_z$\) causes vorticity diffusion.

6. **Conclusion:**
   - The vortex ring deforms into an **elliptical shape** due to stretching.
   - Secondary vorticity introduces **asymmetry** and potential instability.

---

## **Case 2: Vortex Ring in an Expanding Duct**

### **Assumptions:**
- Gradual duct expansion (area \($A_1 \rightarrow A_2$\), \($A_2 > A_1$\)).
- Incompressible flow (\($\nabla \cdot \vec{u} = 0$\)) ⇒ \($u_1 A_1 = u_2 A_2$\).

### **Mathematical Development:**

1. **Axial velocity reduction:**
   - Mass conservation implies:
     \[
     $u_z = \frac{Q}{A(z)} \quad \text{(decreases with expansion)}$
     \]

2. **Initial vorticity (primarily axial):**
   \[
   $\vec{\omega} = \omega_z \hat{e}_z$
   \]

3. **Effect of expansion on vorticity:**
   - **Radial expansion** dilutes vorticity (conservation of circulation \($\Gamma = \oint \vec{u} \cdot d\vec{l}$\)):
     \[
     $\omega_z \propto \frac{1}{r^2} \quad \text{(vorticity weakening)}$
     \]

4. **Dominance of viscous diffusion:**
   - As velocity decreases, \($(\vec{\omega} \cdot \nabla) \vec{u}$\) becomes negligible.
   - **Viscous diffusion** dominates:
     \[
     $\frac{\partial \omega_z}{\partial t} \approx \nu \left( \frac{1}{r} \frac{\partial}{\partial r} \left( r \frac{\partial \omega_z}{\partial r} \right) \right)$
     \]

5. **Approximate solution (Gaussian vorticity profile):**
   - If initially \($\omega_z(r,0) = \omega_0 e^{-r^2 / \sigma^2}$\), diffusion leads to:
     \[
     $\omega_z(r,t) \approx \frac{\omega_0}{1 + 4 \nu t / \sigma^2} e^{-r^2 / (\sigma^2 + 4 \nu t)}$
     \]
     - **Vorticity magnitude decays** as \($\sim \frac{1}{t}$\).
     - **Vortex radius grows** as \($\sqrt{\sigma^2 + 4 \nu t}$\).

6. **Conclusion:**
   - The vortex **expands radially** and **weakens**.
   - **Viscous diffusion** dominates, leading to eventual dissipation.

---

### **Summary of Key Mathematical Effects:**

<center>

| **Case**               | **Dominant Term**           | **Governing Equation**                                                                 | **Approximate Solution**                                                                 |
|------------------------|-----------------------------|---------------------------------------------------------------------------------------|-----------------------------------------------------------------------------------------|
| **Curved Duct (45°)**  | \($(\vec{\omega} \cdot \nabla) \vec{u}$\) | \($\frac{D \omega_z}{Dt} = \omega_z \frac{\partial u_z}{\partial z} + \nu \nabla^2 \omega_z$\) | Elliptical deformation, potential instability.                                          |
| **Expanding Duct**     | \($\nu \nabla^2 \vec{\omega}$\)           | \($\frac{\partial \omega_z}{\partial t} \approx \nu \frac{1}{r} \frac{\partial}{\partial r} \left( r \frac{\partial \omega_z}{\partial r} \right)$\) | \($\omega_z(r,t) \approx \frac{\omega_0}{1 + 4 \nu t / \sigma^2} e^{-r^2 / (\sigma^2 + 4 \nu t)}$\) |

</center>

### **Final Observations:**
- In **curved ducts**, **vortex stretching** dominates, causing deformation and instability.
- In **expanding ducts**, **viscous diffusion** dominates, leading to vortex dissipation.
